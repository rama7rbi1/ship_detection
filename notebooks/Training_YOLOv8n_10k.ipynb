**Load YOLOv8n Model**

```python
model = YOLO('yolov8n.pt')
```

- Baseline

```python
metrics = model.val(data = '/content/drive/MyDrive/ships-aerial-images-updated/data.yaml')
```

- Train the Model

```python
model.train(
    data = '/content/drive/MyDrive/ships-aerial-images-updated/data.yaml',
    epochs = 100,
    imgsz = 640,
    batch = 16,
    name = 'ship_detector_v8n_v3'
)
```

**Plot Training Results**
